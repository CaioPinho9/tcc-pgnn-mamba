# Ajuste de hiperparâmetros do Mamba com Optuna · Marmousi (500 épocas)

```python
from tuning.optuna import (
    create_or_load_study,
    display_trials,
    export_best_trial_params,
    plot_optimization_history,
    plot_param_importances,
    print_search_ranges,
    run_optuna_study,
    setup_optuna_notebook,
)
```

```python
MODEL_TYPE = "mamba"
VALIDATION_DATASET_NAMES = ["Marmousi"]
SAMPLE_INDICES = [0, 1]
DATASETS_ROOT = None  # Defina isto quando o conjunto de dados estiver montado fora do repositório.
MAX_TRIALS = 100
SEARCH_SEED = 24
OBJECTIVE_METRIC = "model_loss"
OBJECTIVE_AGGREGATE = "mean_last_10"

SEARCH_PARAMS = [
    "learning_rate",
    "adam_beta1",
    "adam_beta2",
    "lr_gamma",
    "lr_milestone",
    "gradient_clip_norm",
    "n_blocks_encoder",
    "n_blocks_decoder",
    "use_mamba_encoder",
    "use_mamba_bottleneck",
    "use_mamba_decoder",
    "mamba_base_channels",
    "mamba_d_conv",
    "mamba_d_state",
    "mamba_expand",
    "mamba_dropout",
    "mamba_conv_kernel_size",
    "use_pos_embed",
    "use_skip_connection",
]
optuna_config = setup_optuna_notebook(
    MODEL_TYPE,
    validation_dataset_names=VALIDATION_DATASET_NAMES,
    search_params=SEARCH_PARAMS,
    max_trials=MAX_TRIALS,
    search_seed=SEARCH_SEED,
    objective_metric=OBJECTIVE_METRIC,
    objective_aggregate=OBJECTIVE_AGGREGATE,
    sample_indices=SAMPLE_INDICES,
    datasets_root=DATASETS_ROOT,
    optuna_study_version="marmousi500_fixed_v9",
)
```

## Espaço de busca

```python
print_search_ranges(optuna_config)
```

## Estudo

```python
study, remaining_trials = create_or_load_study(optuna_config)
```

## Execução

```python
run_optuna_study(optuna_config, study, remaining_trials)
```

## Tentativas

```python
display_trials(study)
```

## Importância dos hiperparâmetros

```python
plot_param_importances(optuna_config, study)
```

```python
plot_optimization_history(optuna_config, study)
```

## Exportação

```python
export_best_trial_params(optuna_config, study)
```
